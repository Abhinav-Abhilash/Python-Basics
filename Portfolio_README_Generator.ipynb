```python
# Task 4 - Portfolio README Generator
name = input("Name: ")
role = input("Role: ")
skills = input("Skills: ")
project = input("Project Name: ")
github = input("GitHub Link: ")
linkedin = input("LinkedIn Link: ")

print("\n" + "=" * 45)
print("             README.md CONTENT")
print("=" * 45)
print(f"# {name}")
print("\n## Overview")
print(f"I am a {role} focused on building clean, efficient, and scalable applications.")
print(f"My technical background includes working extensively with {skills}.")
print(f"Currently, I am developing {project}, where I am exploring new ways to solve problems through code.")
print("\n## Let's Connect")
print("I am always open to discussing tech, projects, or new opportunities.")
print(f"- LinkedIn: {linkedin}")
print(f"- GitHub: {github}")
print("=" * 45)
```
